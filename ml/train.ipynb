```python
import numpy as np
import pandas as pd
```

```python
def angle(a, b, c):
    a, b, c = np.array(a), np.array(b), np.array(c) 
    v1 = a - b
    v2 = c - b
    dot = np.dot(v1,v2)
    length_1 = np.linalg.norm(v1)
    length_2 = np.linalg.norm(v2)
    cos = dot / (length_1*length_2)
    clip = np.clip(cos, -1, 1)
    arccos = np.arccos(clip)
    degrees = np.degrees(arccos)
    
    return(np.degrees(np.arccos(clip)))

angle((0,0),(0,10),(10,10))            # 90.0
angle((0,0),(0,10),(0,20))            # 180.0
angle((0,0),(0,10),(10,20))          # 135.0
angle((0.1,0.2),(0.3,0.6),(0.5,1.0))

```
